# Part 2

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode())
df['Age'] = df['Age'].fillna(df['Age'].mean())
agesex_median = df.groupby(['Sex', 'Pclass'])['Age'].transform('median')
df['Age'] = df['Age'].fillna(agesex_median)
df.head(15)
```

# Part 3

```python
FamilySize = df['SibSp'] + df['Parch'] + 1
FamilySize.describe()
```

The mean family size was 1.9 (approximately 2), the minimum family size was 1, and the maximum family size was 11 on the Titanic.

```python
df[FamilySize<=1].count()
```

# Survival Rate by percentage, sex and class
| Sex | Passenger Class 1 | Passenger Class 2 | Passenger Class 3 |
|----------|----------|----------|----------|
| Female  | 96.81%  | 92.11%  | 50.00%  |
| Male  | 36.89%  | 15.74%  | 13.54%  |

Regardless of sex, passenger class 1 had the highest survival rate (e.g. compare 37% to 13-15% survival for males). But females had a higher survival rate overrall, with the difference in highest survival rate being 59.92% (female - male; 96.81% - 36.89%). Thus, you are more likely to survive the Titanic disaster as a female in the first passenger class.
